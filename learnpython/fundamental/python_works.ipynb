Use Case 2: Sum of Even and Odd Numbers

```python
num=int(input("Enter the Number"))
even_sum=0
odd_sum=0
for i in range(1,num+1):
    if i%2==0:
        even_sum=even_sum+i
    else:
        odd_sum=odd_sum+i

print(f"Sum of Even Number:{even_sum}\nSum of Odd Number:{odd_sum}")
```
